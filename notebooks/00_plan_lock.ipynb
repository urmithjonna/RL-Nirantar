# 00 · Lock the final study plan
`results/FINAL_SPEC.md` is hashed and locked here, before any Pune table is built and before any model of this study is trained.
Every later notebook checks this hash first.

```python
import json, hashlib, datetime
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
spec, lock = ROOT / "results" / "FINAL_SPEC.md", ROOT / "results" / "FINAL_SPEC_lock.json"
h = hashlib.sha256(spec.read_bytes()).hexdigest()
if lock.exists():
    assert json.loads(lock.read_text())["sha256"] == h, "FINAL_SPEC.md changed after it was locked"
    print("plan already locked:", json.loads(lock.read_text()))
else:
    lock.write_text(json.dumps({"file": spec.name, "sha256": h, "locked_at": datetime.datetime.now().isoformat(timespec="seconds")}, indent=1))
    print("plan locked:", h[:16])
# reference models copied from the archive (earlier best: PPO-BC v3; its frozen forecasters)
ref = {p.relative_to(ROOT).as_posix(): hashlib.sha256(p.read_bytes()).hexdigest() for p in sorted((ROOT / "models" / "reference").rglob("*")) if p.is_file()}
(ROOT / "results" / "reference_manifest.json").write_text(json.dumps(ref, indent=1))
for k, v in ref.items(): print(v[:16], k)
```
